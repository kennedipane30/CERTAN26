```python
# 1. Membaca dataset
import pandas as pd
df = pd.read_csv('Salary_Data.csv')
print(df.head())
```

Simple Linear Regression

```python
# 2. Memisahkan fitur (X) dan target (y)
X = df.iloc[:, :-1].values
y = df.iloc[:, 1].values
```

```python
# 1. Membaca dataset berganda
df2 = pd.read_csv('multiple_linear_regression_dataset.csv')
print(df2.head())
```

```python
# 3. Visualisasi sebaran data awal
import matplotlib.pyplot as plt
plt.scatter(X, y, color="red")
plt.xlabel('Years Experience')
plt.ylabel('Salary')
plt.show()
```

```python
# 4. Membagi dataset menjadi data latih dan data uji
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
# 5. Melatih model regresi linear
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
# 6. Menentukan intercept dan slope (persamaan garis)
a = regressor.intercept_
b = regressor.coef_[0]
print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"y_hat = {a:.2f} + {b:.2f}x")
```

```python
# 7. Memprediksi data uji
y_pred = regressor.predict(X_test)
print(y_pred)
```

```python
# 8. Mengevaluasi akurasi model
from sklearn.metrics import r2_score
score = r2_score(y_test, y_pred)
print(score)
```

```python
# 9. Visualisasi hasil garis regresi pada data latih
plt.scatter(X_train, y_train, color='red')
plt.plot(X_train, regressor.predict(X_train), color='blue')
plt.show()
```

```python
# 10. Visualisasi hasil garis regresi pada data uji
plt.scatter(X_test, y_test, color='red')
plt.plot(X_train, regressor.predict(X_train), color='blue')
plt.title('Salary vs Experience (Test set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

Multiple Linear Regression

```python
# 2. Memisahkan input (banyak kolom) dan target
X2 = df2.iloc[:, :-1]
y2 = df2.iloc[:, 2]
```

```python
# 3. Membagi dataset menjadi data latih dan uji
X_train2, X_test2, y_train2, y_test2 = train_test_split(X2, y2, test_size=0.2, random_state=0)
```

```python
# 4. Melatih model regresi linier berganda
regressor = LinearRegression()
regressor.fit(X_train2, y_train2)
```

```python
# 5. Memprediksi data uji
y_pred2 = regressor.predict(X_test2)
print(y_pred2)
```

```python
# 6. Menghitung intercept dan koefisien masing-masing fitur
a_ = regressor.intercept_
b1 = regressor.coef_[0]
b2 = regressor.coef_[1]
print("Estimated model intercept, a:", a_)
print("Estimated model slope, b1:", b1)
print("Estimated model slope, b2:", b2)
print(f"y_hat = {a_:.2f} + {b1:.2f}x1 + {b2:.2f}x2")
```

```python
# 7. Evaluasi akurasi model berganda
score2 = r2_score(y_test2, y_pred2)
print(score2)
```
