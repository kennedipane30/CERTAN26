```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

X = df.drop(columns=["Species"])
y = df["Species"]
```

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

memerlukan angka untuk melakukan perhitungan karena komputer kurang memahami teks nama bunga secara langsung. Kode ini menggunakan alat bernama LabelEncoder untuk mengubah nama-nama jenis bunga pada variabel y (seperti setosa, versicolor, dan virginica) menjadi kode angka sederhana, yaitu 0, 1, dan 2.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Membagi data menjadi dua kelompok, yaitu data untuk belajar  dan data untuk ujian . Nilai test_size=0.2 berarti sebanyak 20% data disimpan untuk bahan ujian model, sedangkan 80% sisanya dipakai untuk proses pembelajaran, dengan pembagian baris yang dilakukan secara acak namun konsisten.

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

 pembuatan model kecerdasan buatannya. Model yang dipilih adalah Gaussian Naïve Bayes karena seluruh data ukuran bunga berupa angka desimal berkelanjutan . Lewat perintah .fit(), model mulai "belajar" menghubungkan antara ukuran ciri-ciri bunga pada data latih dengan jenis bunganya masing-masing

```python
result = NB_model.predict(X_test)
result
```

 menguji kemampuannya menggunakan fungsi .predict() pada data uji yang belum pernah ia lihat sebelumnya. Model mencoba menebak jenis bunga berdasarkan ukuran yang diberikan, dan hasilnya disimpan dalam variabel result berupa deretan kode angka tebakannya.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

menghitung nilai ujian atau tingkat ketepatan tebakan model. Fungsi accuracy_score akan mencocokkan hasil tebakan model dengan jenis bunga asli yang sebenarnya, lalu mengalikannya dengan angka 100 untuk menampilkan nilai akurasi dalam bentuk persentase.

```python
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

mencoba memberikan satu sampel ukuran bunga baru dalam bentuk daftar angka biasa untuk ditebak oleh model. Hasil tebakan angka dari model kemudian diterjemahkan kembali ke nama asli bunganya menggunakan inverse_transform, meski cara memasukkan data ini memunculkan pesan peringatan karena formatnya belum menyertakan nama kolom ciri-ciri ukuran.
